```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the data we scraped earlier
df = pd.read_csv('clean_books_data.csv')

# 2. EDA: Basic Information
print("--- First 5 Rows ---")
print(df.head())

print("\n--- Dataset Info ---")
print(df.info())

print("\n--- Statistical Summary ---")
print(df.describe())

# 3. EDA: Asking Questions
# What is the average price?
avg_price = df['Price'].mean()
print(f"\nThe average price of a book is: £{avg_price:.2f}")

# How many books are in each rating?
print("\n--- Rating Counts ---")
print(df['Rating'].value_counts().sort_index())
```

```python
# Set the visual style for our charts
sns.set_theme(style="whitegrid")

# Create a figure with 3 subplots (1 row, 3 columns)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# --- Chart 1: Count of Books by Rating ---
sns.countplot(data=df, x='Rating', palette='viridis', ax=axes[0])
axes[0].set_title('Number of Books per Rating')
axes[0].set_xlabel('Star Rating')
axes[0].set_ylabel('Count')

# --- Chart 2: Distribution of Book Prices ---
sns.histplot(df['Price'], bins=15, kde=True, color='blue', ax=axes[1])
axes[1].set_title('Distribution of Book Prices')
axes[1].set_xlabel('Price (£)')
axes[1].set_ylabel('Frequency')

# --- Chart 3: Price vs Rating ---
sns.boxplot(data=df, x='Rating', y='Price', palette='Set2', ax=axes[2])
axes[2].set_title('Price Distribution by Rating')
axes[2].set_xlabel('Star Rating')
axes[2].set_ylabel('Price (£)')

# Adjust layout so titles don't overlap
plt.tight_layout()

# Save the figure to use in your GitHub/README
plt.savefig('books_visualizations.png')

# Show the charts
plt.show()
```
